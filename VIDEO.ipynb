```python
import torch
from diffusers import DiffusionPipeline
from diffusers.utils import export_to_video

# 1. Load the Text-to-Video Diffusion Pipeline
print("Loading model... (this may take 1-2 minutes)")
pipe = DiffusionPipeline.from_pretrained(
    "damo-vilab/text-to-video-ms-1.7b",
    torch_dtype=torch.float16,
    variant="fp16"
)

# Enable memory optimizations to prevent running out of GPU VRAM
pipe.enable_model_cpu_offload()

# 2. Define your prompt and generate video frames
prompt = "An astronaut walking on the surface of Mars, dramatic lighting, photorealistic, 4k"
print(f"Generating video for prompt: '{prompt}'")

# num_frames controls video length (16 frames ~ 2 seconds)
video_frames = pipe(prompt, num_frames=16, num_inference_steps=25).frames[0]

# 3. Save generated frames to an MP4 video file
output_video_path = export_to_video(video_frames, "generated_video.mp4")
print(f"Video saved to: {output_video_path}")
```

```python
from IPython.display import HTML
from base64 import b64encode

# Read and encode the video file
mp4_file = open('generated_video.mp4', 'rb').read()
data_url = "data:video/mp4;base64," + b64encode(mp4_file).decode()

# Render inline HTML video player
HTML(f'''
<video width="400" height="400" controls autoplay loop>
    <source src="{data_url}" type="video/mp4">
</video>
''')
```
