```python
!pip install -U aiogram
!pip install openai
import asyncio
from aiogram import Bot, Dispatcher, types
import os
import sys
import logging
from google.colab import userdata

os.environ["TELEGRAM_BOT_TOKEN"] = userdata.get('TELEGRAM_BOT_TOKEN')
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')

# Initialize bot and dispatcher
bot = Bot(token=os.environ["TELEGRAM_BOT_TOKEN"])
dp = Dispatcher()
```

```python
# now adding multiple commands matlab kis key seh kya hota hai
from aiogram.filters import Command
@dp.message(Command('start'))
async def command_start(message: types.Message):
  # this handler will recieve the message  with /start command
    await message.reply('hi!\n I amd working as a bot')
```

```python
# matlab jab user /help press kre tabh kya krna hai
@dp.message(Command('help'))
async def command_help(message: types.Message):
  help_text='Reffer to these commands'
  await message.reply(help_text)
```

```python
#using chatgpt
@dp.message()
async def other_messages(message: types.Message):
  "Will handle all other messages"

  print(f"User: {message.text}")

  context.append({"role": "user", "content": message.text})

  response = openai.ChatCompletion.create(
      model="gpt-4o-mini",
      messages=context
  )

  Value = response.choices[0].message.content
  print(f"Bot response: {value}")

  print(f"Bot: {Value}")

  context.append({"role": "assistant", "content": Value})

  await message.reply(Value)
```

```python
# starting bot
async def main() -> None:
  bot = Bot(token=os.environ["TELEGRAM_BOT_TOKEN"])
  await dp.start_polling(bot)
```

```python
if __name__ == '__main__':
  logging.basicConfig(leel=logging.INFO, stream=sys.stdout)
  await main()
```
